```python
# === 模块一：变量和数据类型 ===

# --- 整数 int ---
spin_speed = 3200          # 纺丝速度，单位 m/min，这是一个整数
bobbins_per_shift = 48     # 每班落筒数量

# --- 浮点数 float（带小数点的数） ---
tenacity = 2.85            # 断裂强度，单位 cN/dtex
elongation = 32.4          # 断裂伸长率，单位 %
temp_C = 285.0             # 纺丝箱体温度，单位 °C

# --- 字符串 str（文本，用引号括起来） ---
product_code = "FDY-75-36" # 品种代码：75旦/36孔 FDY
luster = "半光"             # 光泽类型

# --- 布尔值 bool（只有两个值：True 或 False） ---
is_qualified = True        # 这批产品是否合格
alarm_on = False           # 报警是否触发

# --- 用 print() 输出，用 type() 查看类型 ---
print(spin_speed)          # 输出：3200
print(type(spin_speed))    # 输出：<class 'int'>
print(type(tenacity))      # 输出：<class 'float'>
print(type(product_code))  # 输出：<class 'str'>
print(type(is_qualified))  # 输出：<class 'bool'>

# --- 字符串拼接 ---
shift = "A班"
message = shift + " 纺速：" + str(spin_speed) + " m/min"
print(message)             # 输出：A班 纺速：3200 m/min

# --- f-string：更现代的拼接写法（推荐用这个） ---
message2 = f"{shift} 纺速：{spin_speed} m/min，断裂强度：{tenacity} cN/dtex"
print(message2)
```

```python
# === 模块二：列表和字典 ===

# --- 列表 list ---
tenacity_log = [2.81, 2.85, 2.79, 2.90, 2.83, 2.77]  # 某班6次断裂强度测量值

print(tenacity_log[0])      # 输出第1个：2.81（索引从0开始）
print(tenacity_log[-1])     # 输出最后一个：2.77（-1就是倒数第一）
print(tenacity_log[1:4])    # 输出第2到第4个：[2.85, 2.79, 2.90]（切片，不含右端）

tenacity_log.append(2.88)   # 在末尾添加一个新数据
print(len(tenacity_log))    # 输出列表长度：7

print(max(tenacity_log))    # 输出最大值：2.90
print(min(tenacity_log))    # 输出最小值：2.77

# --- 字典 dict ---
batch_record = {
    "品种":    "FDY-75-36",
    "纺速":    3200,
    "温度":    285.0,
    "断裂强度": 2.85,
    "合格":    True
}

print(batch_record["品种"])      # 用键取值，输出：FDY-75-36
print(batch_record["断裂强度"])  # 输出：2.85

batch_record["伸长率"] = 32.4    # 新增一个键值对
batch_record["纺速"] = 3300      # 修改已有的值

print(batch_record)
print(batch_record.keys())       # 输出所有键
print(batch_record.values())     # 输出所有值

# --- 列表里装字典：表格的原始形态 ---
production_log = [
    {"品种": "FDY-75-36", "纺速": 3200, "断裂强度": 2.85},
    {"品种": "FDY-100-36", "纺速": 2900, "断裂强度": 3.12},
    {"品种": "POY-150-48", "纺速": 3800, "断裂强度": 1.92},
]

print(production_log[1])              # 取第2条记录（整个字典）
print(production_log[1]["断裂强度"])  # 取第2条记录的断裂强度
```
